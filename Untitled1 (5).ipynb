```python
from sentence_transformers import SentenceTransformer
from transformers import pipeline
import numpy as np
import faiss

print("FAISS version:", faiss.__version__)

# --------------------------------------------------
# 1. Knowledge Base
# --------------------------------------------------

documents = [
    "The Eiffel Tower is located in Paris, France and was completed in 1889.",
    "Retrieval-Augmented Generation combines document retrieval with text generation.",
    "Python is a popular high-level programming language used in AI development.",
    "Vector databases store embeddings and support fast similarity search."
]

# --------------------------------------------------
# 2. Load Embedding Model
# --------------------------------------------------

print("\nLoading embedding model...")

embed_model = SentenceTransformer("all-MiniLM-L6-v2")

# Convert documents into embeddings
doc_embeddings = embed_model.encode(
    documents,
    convert_to_numpy=True
).astype("float32")

print("Embedding shape:", doc_embeddings.shape)

# --------------------------------------------------
# 3. Build FAISS Index
# --------------------------------------------------

dimension = doc_embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

# Add document embeddings to FAISS
index.add(doc_embeddings)

print("Documents stored in FAISS:", index.ntotal)

# --------------------------------------------------
# 4. User Query
# --------------------------------------------------

query = "What is RAG in AI?"

# Convert query into embedding
query_embedding = embed_model.encode(
    [query],
    convert_to_numpy=True
).astype("float32")

# --------------------------------------------------
# 5. Retrieve Top-2 Relevant Documents
# --------------------------------------------------

distances, indices = index.search(
    query_embedding,
    k=2
)

retrieved_chunks = [
    documents[i]
    for i in indices[0]
]

print("\nQuery:")
print(query)

print("\nRetrieved Context:")

for i, chunk in enumerate(retrieved_chunks, 1):
    print(f"{i}. {chunk}")

# --------------------------------------------------
# 6. Create Augmented Prompt
# --------------------------------------------------

context = " ".join(retrieved_chunks)

prompt = f"""
Answer the question using only the provided context.

Context:
{context}

Question:
{query}

Answer:
"""

# --------------------------------------------------
# 7. Load FLAN-T5
# --------------------------------------------------

print("\nLoading generation model...")

generator = pipeline(
    "text2text-generation",
    model="google/flan-t5-base"
)

# --------------------------------------------------
# 8. Generate Answer
# --------------------------------------------------

answer = generator(
    prompt,
    max_new_tokens=60,
    do_sample=False
)

# --------------------------------------------------
# 9. Display Final Answer
# --------------------------------------------------

print("\nFinal Answer:")
print(answer[0]["generated_text"])
```
